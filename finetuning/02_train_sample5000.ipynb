## 1. Locate the prepared dataset from notebook 1

`INPUT_DATA_DIR` below already points at `visionassist-01-dataset-prep-eda`'s slug (this repo's `kaggle kernels push` setup, `finetuning/kaggle/`, keeps these in sync automatically). If you instead uploaded notebook 1's output manually and Kaggle gave it a different name, check the Data panel on the right and update the path below to match. **Check cell 1's `/kaggle/input` listing above before trusting this path** — `finetuning/kaggle/RUN_LOG.md` (notebook 1, attempts 7-9) found that API-attached *datasets* (`dataset_sources`) mount at `/kaggle/input/datasets/<owner>/<slug>/`, not the flatter `/kaggle/input/<slug>/` assumed below — `kernel_sources` (another kernel's *output*, which is what this is) may follow the same deeper convention, or a different one entirely; this hasn't been verified yet since this notebook has never actually been run. If `INPUT_DATA_DIR` below is wrong, cell 1's listing tells you the real path.

```python
from pathlib import Path

# Kaggle's kernel_sources mount path is NOT stable across pushes — confirmed live on
# the freeze=None sweep variant (same single kernel_sources shape, mounted under
# /kaggle/input/notebooks/<owner>/<slug>/ instead of a flat /kaggle/input/<slug>/).
# Search for the file instead of hardcoding either convention.
def _find_mounted_file(filename: str, parent_dir_name: str) -> Path:
    matches = [p for p in Path("/kaggle/input").rglob(filename) if p.parent.name == parent_dir_name]
    assert matches, f"No '{filename}' found under a '{parent_dir_name}/' directory anywhere under /kaggle/input — check cell 1's listing above and that the input dataset is actually attached."
    assert len(matches) == 1, f"Multiple '{filename}' matches under '{parent_dir_name}/': {matches} — can't tell which is the real one."
    return matches[0]

DATA_YAML = _find_mounted_file("lostfound.yaml", "lostfound")
print(f"Found dataset yaml at: {DATA_YAML}")
```

## 2. Checkpoint / resume

Set `RESUME = True` only if you're continuing a previous attempt that got cut off. If so, **Add Data** → your own previous `02_train.ipynb` run's output version (the one you Saved right before the session ended) as a *second* input, then point `PREVIOUS_RUN_DIR` at it below. This copies that run's `weights/last.pt` + training state into place and resumes from the last completed epoch — it does **not** restart from `yolo26n.pt`.

```python
RESUME = False  # <-- set True to continue a previous cut-off attempt
PREVIOUS_RUN_DIR = Path("/kaggle/input/<notebook-02-previous-attempt-slug>/runs/lostfound_yolo26n")  # <-- EDIT if RESUME=True

RUN_PROJECT = "/kaggle/working/runs"
RUN_NAME = "lostfound_yolo26n"
RUN_DIR = Path(RUN_PROJECT) / RUN_NAME
LAST_CKPT = RUN_DIR / "weights" / "last.pt"

if RESUME:
    import shutil
    assert PREVIOUS_RUN_DIR.exists(), f"{PREVIOUS_RUN_DIR} not found — attach the previous run's output as an input dataset first."
    RUN_DIR.parent.mkdir(parents=True, exist_ok=True)
    shutil.copytree(PREVIOUS_RUN_DIR, RUN_DIR, dirs_exist_ok=True)
    if not LAST_CKPT.exists():
        raise SystemExit(
            f"RESUME=True but {LAST_CKPT} doesn't exist after copying {PREVIOUS_RUN_DIR}. "
            "The previous attempt likely never completed a single epoch (killed before its "
            "first checkpoint save), or the wrong output version is attached. Fix "
            "PREVIOUS_RUN_DIR above, or set RESUME=False to start fresh instead. Do NOT "
            "ignore this and re-run with RESUME=True anyway — the training cell below has "
            "no fallback, and would otherwise need to fail loudly here rather than silently "
            "starting a fresh run inside this now-partially-populated RUN_DIR, which would "
            "mix stale args.yaml/results.csv from the old attempt with the new run."
        )
    print(f"Copied previous run into {RUN_DIR}. Resuming from {LAST_CKPT}.")
```

## 3. Train

`freeze=10` — see `docs/FINE_TUNING_YOLO_GUIDE.md` Section 4 for the exact frozen/trainable parameter breakdown (1,115,744 frozen / 1,456,536 trainable, 56.62% of the model trains) and Section 5 for why this is full fine-tuning with a frozen backbone, not LoRA/QLoRA.

```python
from ultralytics import YOLO
import torch

# Nothing else in this notebook checks that a GPU is actually in use before this
# expensive call. If GPU provisioning silently degrades (e.g. the free-tier ~30
# GPU-hr/week quota is exhausted) while enable_gpu=true is still honored at
# scheduling, epochs=100 could silently run on CPU for the full schedule — same bug
# class as notebook 1's unattended CPU-epoch incident, just about how fast training
# runs rather than whether it runs at all.
assert torch.cuda.is_available(), "No GPU available — check Settings → Accelerator is set to GPU T4 x2/P100 before running this cell."

# RESUME=True is only ever reached here once cell 5 above has already confirmed
# LAST_CKPT exists (it raises SystemExit otherwise) — no silent fresh-training
# fallback, so a missing checkpoint can never be mistaken for a real resume.
if RESUME:
    print(f"Resuming from {LAST_CKPT}")
    model = YOLO(str(LAST_CKPT))
    results = model.train(resume=True)
else:
    print("Starting a fresh training run.")
    model = YOLO("yolo26n.pt")
    results = model.train(
        data=str(DATA_YAML),
        epochs=100,
        imgsz=640,
        freeze=10,
        optimizer="AdamW",
        lr0=0.001,
        patience=15,
        # notebook 1's EDA found a 5702.9x max/min instance imbalance across the 48
        # classes (5 classes under the 50-instance floor, worst at 14 instances).
        # cls_pw computes inverse-class-frequency weights straight from this run's
        # own training label counts and applies them to the classification loss
        # (ultralytics/models/yolo/detect/train.py's compute_class_weights() —
        # verified directly against the pinned 8.4.106 wheel; default is 0.0,
        # i.e. off). 0.5 dampens rather than fully inverse-weights (1.0) — at this
        # ratio, full inverse weighting risks destabilizing the common classes for
        # a handful of classes that have too few real examples to learn well no
        # matter how they're weighted. Free to enable: it only reweights an
        # existing loss term, no extra epochs or data, so there's no reason to
        # run once without it and again with it.
        cls_pw=0.5,
        project=RUN_PROJECT,
        name=RUN_NAME,
        exist_ok=True,
    )
```

## If the session is about to run out (Kaggle session-limit runbook)

Ultralytics writes `weights/last.pt` after every completed epoch, so progress isn't lost just because the clock is running low — but it only survives a session restart if you commit it:

1. **Interrupt the training cell** (Stop button) — don't let Kaggle kill it for you. Whatever epoch last finished is already safely on disk.
2. **File → Save Version** — commits `/kaggle/working` (including `runs/lostfound_yolo26n/`) as this notebook's Output, which becomes a Kaggle Dataset you can reattach.
3. Next session: open a fresh copy of this notebook, **Add Data** → that previous output version, set `RESUME = True` and `PREVIOUS_RUN_DIR` in Section 2 above, re-run from the top. `resume=True` continues from `last.pt`'s saved epoch + optimizer state, not from scratch.
